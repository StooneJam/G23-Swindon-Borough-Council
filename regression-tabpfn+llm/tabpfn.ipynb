# TabPFN v3

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
import numpy as np
import pandas as pd

train = pd.read_csv("../data preprocessing+EDA/train_updated.csv")
test  = pd.read_csv("../data preprocessing+EDA/test_updated.csv")

feats = ["Urban_rura_Urban",
         "LU_pct_micro_2025_msoa",
         "LU_pct_large_2025_msoa",
         "log_enterprises_per_1k_residents_2025",
         "turnover_diversity_1-HHI_2025_msoa",
         "LU_diversity_1-HHI_2025",
         "share_enterprises_kibs_2025_msoa",
         "emp_rate",
         "full_time_share",
         "log_Mid-2024 population",
         "IMD_decile"]

X_tr, X_te = train[feats].values, test[feats].values

print("train", train.shape)
print("test", test.shape)
```

```python
from tabpfn import TabPFNRegressor
from tabpfn.constants import ModelVersion
from sklearn.metrics import mean_squared_error, mean_absolute_error, mean_absolute_percentage_error, r2_score

reg = TabPFNRegressor.create_default_for_version(ModelVersion.V3)
```

```python
target = "log_gva_per_worker_2023"
reg.fit(X_tr, train[target].values)
pred_reg = reg.predict(X_te)
y = test[target].values

print(f"target: {target}")
print(f"MAE = {mean_absolute_error(y, pred_reg):.4f}")
print(f"RMSE = {np.sqrt(mean_squared_error(y, pred_reg)):.4f}")
print(f"MAPE = {mean_absolute_percentage_error(y, pred_reg):.4f}")
print(f"R2 = {r2_score(y, pred_reg):.4f}")
```

```python
target = "log_total_GVA_2023"
reg.fit(X_tr, train[target].values)
pred_reg = reg.predict(X_te)
y = test[target].values

print(f"target: {target}")
print(f"MAE = {mean_absolute_error(y, pred_reg):.4f}")
print(f"RMSE = {np.sqrt(mean_squared_error(y, pred_reg)):.4f}")
print(f"MAPE = {mean_absolute_percentage_error(y, pred_reg):.4f}")
print(f"R2 = {r2_score(y, pred_reg):.4f}")
```
